***Project title*** : Study of Netflix shows' popularity using IMDB measurement

dataset.png

Netflix has become one of the dominant content platforms. Understanding the factors that influence the popularity of shows is a topic of interest, not only for content creators but also for the audience itself. As a movie watcher, one might decide to pick a good show depend on its popularity, one that capture more attention.

IMDb votes serve as a widely recognized measure of a show's popularity, reflecting the audience's engagement and overall sentiment. This study aims to select several factors and analyze whether those factors might influence the number of IMDb votes for Netflix shows, with a focus on understanding what makes certain titles more popular than others.

***Topic and problem statement***


This study investigates factors influencing the popularity of Netflix shows by analyzing IMDb votes as popularity measurement. To examine whether there are statistically significant differences in IMDb votes between types of Netflix shows (movies and TV series) and among different genres.

***Objectives***

* To evaluate whether types of shows differ significantly in IMDb votes.
* To determine whether genres significantly influence the number of IMDb votes.

***Research design and methodology***

1. Data collection and preprocessing

2. Exploratory Data Analysis
       - We set hypotheses for three factors including, type of  shows, genre popularity, and rating of the shows to differ significantly in number of votes
3. Statistical Test
      - Independent two sample T-test : to compare the number of IMDb votes between movies and TV series
      -One-way Anova: to test if the average number of IMDb votes differed significantly across genres
4. Visualization

***Results and discussion***

This study explored the factors influencing movie popularity, measured as the number of IMDb votes. Results revealed significant differences in popularity among type of shows and genres.

* Type of show- Movies received higher number of votes compared to TV shows.
* Genres- Action genre has the highest average IMDb votes while the sport genre has the lowest one.

From the results, we gained insights of which factors that could increase the popularity. Future research could incorporate other predictors, such as marketing budget or streaming availability, and machine learning techniques, such as simple regression or classification, to provide a more comprehensive understanding and deeper analysis.
